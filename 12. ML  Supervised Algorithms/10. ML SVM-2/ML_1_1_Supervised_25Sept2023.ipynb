## Recap

We learnt the loss function / optimization equation for Hard?Soft margin SVM.

<center><img src='https://drive.google.com/uc?id=1uttM7ibIYdc4Dkr9W4AGVg68ROX6d1yw' width=800></center>



Note that,
- This optimization problem is termed as "**primal form**".

\
When we solve this primal form equation,
- we get optimal $w^*, b^*$


In optimization theory,
- there is a concept called **Primal Dual Equivalance**

which states that
    
```
For a primal form optimization problem,
there exists a dual form optimization problem.
```

and by solving the dual form,
- you'll arrive at **same optimal solution** as primal form.

\
However, </br>
the proof for converting this primal form to dual form is
- challenging and math intensive
- beyond the scope of this course.

## Kernel SVM

Recall that in dual form of SVM,
- we saw that $x_i, x_j$ always occur in pair.


This pair of $x_i^T.x_j$ represents the similarity.
- i.e. finding the similarity between two vectors


<center><img src='https://drive.google.com/uc?id=1LuqhfpdcO3ZfCoQFAQ-8WyltqWQrv70h' width=700></center>




Instead of calculating similarity using dot product,
- we can have our own custom similarity function called **Kernel**


#### How do we use this kernel ?

We simply replace the dot product with the kernel function


<center><img src='https://drive.google.com/uc?id=1Uxpi2rpZNCGbO2UCuRU_eGBJpN8-qEzS' width=700></center>



### But, what's the importance of this Kernel function ?

Say, we have a usecase where
- you have been provided with Amazon **customer data**

But, **there is a catch**
- you **don't have vector representation** $x_i$ available

<br>

All you have is **similarity data** i.e. how similar is a customer to another one.

#### Can we use logistic Regression to classify these customers ?

**No**, since we do not have any other vector representation $x_i$
- and in the optimization problem, **logLoss needs $x_i$**




but we can **surely use SVM model in dual form**
- as we don't explicitly need $x_i's$ in dual form
- as long as similiarty data is available.



### But, how is all this helping in classifying non linear data ?

Let's look into that.

Say, we have a data distribution which looks like :


<center><img src='https://drive.google.com/uc?id=1zKybit2EqJEkmXMX3Ywnz_TXkNX1NktY' width=700></center>





#### Can we use logistic regression to classify this ?

Simple logistic regression will fail to classify concentric disk data.
- as it'll always give a linear line.


However, by using feature engineering
- i.e. higher degree polynomial $f_1^2, f_2^2$
- we'll be able to classify it.

Do NOTE:
- Here, we are taking help of manual feature engineering
    - in order to classify the data

There's no automated feature engineering here.

However, in SVM,

- there is something called **kernelization**

In simple terms,
- kernelization takes data from d dimension to d' dimension
- where d' >> d

Taking data to higher dimension makes data linear separable.
- Hence, making it easy to classify.


Let's see how it does that.

Specifically, let's look into Polynomial Kernel

### Polynomial Kernel

Polynomial Kernel is written as :


<center><img src='https://drive.google.com/uc?id=1YUXlYQ0UjnT7pUbrTFwMUFW7SievcR6F' width=700></center>



Here, m is the degree
- when m = 2, it becomes quadratic kernel
- when m = 3, it becomes cubic kernel.




Let's write the equation for quadratic kernel.

For simplicity,
- assume c = 1.


<center><img src='https://drive.google.com/uc?id=1wbyCwu9Mo-FnwFDCNCIBCvfkz0aaQlVY' width=700></center>



Say, we have two datapoints $x_1, x_2$ in 2 - dim setup such that:

$x_1 = [x_{11},x_{12}]$ and $x_2 = [x_{21},x_{22}]$

If we were to expand the equation, it'll come out to be.


<center><img src='https://drive.google.com/uc?id=16RtXQfmXtZpVJ4TDxqXIST5soYn6MkF7' width=700></center>



$ $
Supposedly, we have two 6 dimensional vector $x_1', x_2'$ such that:
- $x_1' = [1, x_{11}^2, x_{12}^2, \sqrt{2} x_{11},  \sqrt{2} x_{12},  \sqrt{2} x_{11}x_{12} ]$

- $  x_2' = [ 1, x_{21}^2, x_{22}^2, \sqrt{2} x_{21},  \sqrt{2} x_{22},  \sqrt{2} x_{21}x_{22} ] $

<br>

Now, if we perform dot product of $x_1'$  and $x_2'$, we will find that
- $k(x_1,x_2) = (x_1'^T x_2')$

Notice that
- Both $x_1'$ and $x_2'$ are 6- dim vectors
- and their dot product is giving the same result as what we got from quadratic kernel





This means
- quadratic kernel outcome is result of 6 - dim dot product
- but we were able to get that using 2 - dim vector only ($x_1, x_2$)

#### What does it signify ?

**Originally**, we are in **2-dim space** but
- the **quadratic kernel** has **implicitly** **transformed** it into **6 - dim space**


<center><img src='https://drive.google.com/uc?id=16-T1TY146NAjL1_1-yruwGdiMgyFzvzR' width=700></center>



#### How is it helping ?

If we were to use this quadratic kernel on the previous given non linear data,
- we don't need to do manual feature engineering

as the kernel trick will implicitly project the data to higher dimension
- where data will become separable.



In simple terms,
- while being the 2 dim space
    - we are getting the result of 6 - dim space without explicitly adding features.


This is true power of **Kernelization**

<br>

We are not exactly finding hyperplane in d' dimension

- but solving kernel function is equivalent to going in d' and finding a hyperplane


hence we said,

- Kernel function tricks by projecting data from d dimension to d' dimension **implicitly**

#### How does SVM works with RBF Kernel ?

We simply replace
- $ x_i^Tx_j $ with RBF Kernel $ K_{RBF}(x_i,x_j)$

<br>

Now suppose, we have some datapoints such that:




<center><img src='https://drive.google.com/uc?id=1bjuR3HpYRWwLlt7MlFZdkG7q9o8MJq-J' width=500></center>




Notice that
- we can't classify these points using linear classifier

Let's see how SVM with RBF kernel will help us here

#### How'll RBF kernel classify this ?



Recall when applying kernel functions,
- we transform the d dimensional data to d' dimensional data



Consider two points $x_1$ and $x_2$
 - Since both the points are close to each other,
 - the distance $d_{12}$ will be close to 0,

thus the Kernel value will be high for them







<center><img src='https://drive.google.com/uc?id=1ge4tDWQpRjDkv6c7SFS_Zx1zFM9ANzHf' width=700></center>





Hence these datapoints in d' dimension ($x_1'$ and $x_2'$ ) - will be very close to each other as well.




<center><img src='https://drive.google.com/uc?id=1KALRgIaoUa0ifGlUhUiZXXym6usMNdUt' width=700></center>



Similarly,
- $ x_1 and x_3 $ have a high euclidean distance   ($d_{13}$)

<br>

Hence the kernel value will be exponentially small (for small σ value),  
 - Thus making $x_1'^T x_3' $ to be small

<br>

Hence making the points ($x_1'$ and $x_3'$) far away from each other in d' dimension



#### To summarize:

We have a some +ve and -ve datapoints in d dimension
- For **small value of $ \sigma $**, in d' dimension,
  - **RBF kernel** will be **grouping datapoints** **which are close to each other**


This is similar to what we learnt in KNN

Recall that
- for a small value of K in KNN model
    - will also be making the same grouping of data.

## Code Walkthrough

````python
from matplotlib import pyplot as plt
import numpy as np
from sklearn.inspection import DecisionBoundaryDisplay
````

````python
# create a non-linear dataset

from sklearn.datasets import make_circles
X, y = make_circles(100, factor=.1, noise=.1, random_state=0)
plt.scatter(X[:, 0], X[:, 1], c=y)
plt.show()
````

````python
np.unique(y,return_counts=True)
````

### Linear SVM

SVM sklearn doc: https://scikit-learn.org/stable/modules/generated/sklearn.svm.SVC.html#sklearn-svm-svc

````python
from sklearn.svm import SVC
````

````python
# try using svm with linear kernel
clf = SVC(kernel='linear').fit(X, y)
````

````python
# plot the decision function

plt.scatter(X[:, 0], X[:, 1], c=y, s=50)
plt.title("SVM with linear kernel on non-linear dataset")

ax = plt.gca()
DecisionBoundaryDisplay.from_estimator(
    clf,
    X,
    plot_method="contour",
    colors="k",
    levels=[-1, 0, 1],
    alpha=0.5,
    linestyles=["--", "-", "--"],
    ax=ax,
)
# plot support vectors
ax.scatter(
    clf.support_vectors_[:, 0],
    clf.support_vectors_[:, 1],
    s=100,
    linewidth=1,
    facecolors="none",
    edgecolors="k",
)
plt.show()
````

### RBF kernel

````python
clf = SVC(kernel='rbf', C=1E0) # C = 1E6
clf.fit(X, y)
````

````python
#plot RBF kernel

plt.scatter(X[:, 0], X[:, 1], c=y, s=50)
plt.title("SVM with RBF kernel on non-linear dataset")

ax = plt.gca()
DecisionBoundaryDisplay.from_estimator(
    clf,
    X,
    plot_method="contour",
    colors="k",
    levels=[-1, 0, 1],
    alpha=0.5,
    linestyles=["--", "-", "--"],
    ax=ax,
)
# plot support vectors
ax.scatter(
    clf.support_vectors_[:, 0],
    clf.support_vectors_[:, 1],
    s=100,
    linewidth=1,
    facecolors="none",
    edgecolors="k",
)
plt.show()
````

## Will outliers effect SVM models too?



If we are using SVM without any kernels
- the hyperplane only depends upon support vectors
- so, outliers will not impact the hyperplane

\
However,
- As kernel SVMs are based on distance/proximity and similarlity
- so, outliers will impact them

## Why is SVM not used much?



One of the biggest problem with SVM is -
- its train time complexity is very high
- order of square of the number of datapoints
- i.e. $O(n^2)$

\
Also the run time complexity of SVM
- depends on number of support vectors
- which can be very high or very low

\
Hence, we have no control over the number of Support Vectors.
