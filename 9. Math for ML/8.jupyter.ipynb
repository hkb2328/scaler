```python
import numpy as np
import itertools as it
import matplotlib.pyplot as plt
```

```python
def get_partials(func, x, y):
    delta = 0.001
    fx = (func(x + delta, y) - func(x, y)) / delta
    fy = (func(x, y + delta) - func(x, y)) / delta
    return fx, fy
```

```python
get_partials( lambda x,y : x**2+y**2, 2 , 3)
```

```python
lim = 10 
x = np.linspace(-lim, lim, 20)
y = np.linspace(-lim, lim, 20)
```

```python
Output = np.array(list(it.product(x,y)))
```

```python
u , v = get_partials( lambda x,y : x**2+ 0.00001*y**2, Output[:,0],Output[:,1] )
```

```python
def plot_gradient(g1, g2, U, V):
    fig, ax = plt.subplots()
    q = ax.quiver(g1, g2, U, V)
    plt.show()
```

```python
plot_gradient(Output[:,0],Output[:,1], u, v )
```
