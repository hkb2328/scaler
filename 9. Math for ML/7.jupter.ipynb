```python
import numpy as np 
import matplotlib.pyplot as plt
```

```python
def plot_function(func):
    x_values = np.linspace(-10, 10, 1000)
    y_values = [func(x) for x in x_values]
    plt.plot(x_values, y_values)
```

```python
plot_function(lambda x: x**2)
```

```python
def plot_derivative(func):
    x_values = np.linspace(-10, 10, 1000)
    delta_x = 0.0001
    derivative = (func(x_values + delta_x) - func(x_values))/delta_x
    plt.plot(x_values, derivative)
```

```python
plot_derivative(lambda x: 2*x**2 + 3*x + 1)
```

```python
plot_derivative(lambda x: 1 + np.log(np.exp(x)))
```

```python
def gradient_descent(func, x0):
    list_of_points = []
    x = x0
    delta = 0.0001
    for i in range(100):
        derivative = (func(x + delta) - func(x))/delta
        x = x - 0.1*derivative # learning rate needs to be chosen carefully
        list_of_points.append(x)
    return list_of_points
```
