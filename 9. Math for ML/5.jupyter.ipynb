```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
from sklearn import datasets
```

```python
x, y = datasets.make_blobs(n_samples=200, centers=[[1, 2], [7, 8]], n_features=2, center_box=(0, 10))
```

```python
plt.scatter(x[:, 0], x[:, 1], c=y)
plt.show()
```

```python
y
```

```python
y[y==0] = -1
```

```python

def show(w, w0, x, y):
    line_x = np.linspace(-5, 10)
    line_y = (-w[0]*line_x - w0)/w[1] 
    plt.plot(line_x, line_y)                        # show h-plane
    plt.scatter(x[:, 0], x[:, 1], c=y)
    plt.show()
```

```python
w = np.random.normal(size=(2,))
w0 = np.random.normal()
```

```python
print(w,w0)
```

```python
show(w, w0, x, y)
```

```python
def algorithm(x, y, w, w0, num_iterations=100):
    for iteration in range(num_iterations):
        miss_cls = 0
        for point in range(len(x)):
            hs = np.sign(np.dot(w, x[point]) + w0)
            if hs != np.sign(y[point]):
                miss_cls += 1
                w = w+x[point]*np.sign(y[point]) # courtesy Sintu
                w0 += np.sign(y[point])
                break # one misclassification at a time
        if miss_cls == 0:
            show(w, w0, x, y)
            return w, w0
    show(w, w0, x, y)
    return w, w0
```

```python
show(w, w0, x, y)
```

```python
w, w0 = algorithm(x, y, w, w0, num_iterations=10)
```

```python
print(w,w0)
```
