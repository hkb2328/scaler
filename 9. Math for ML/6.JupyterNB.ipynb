```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
def function_1(x):
    if x <3:
        return 2 
    else:
        return 5 
```

```python
function_1(2)
```

```python
function_1(3)
```

```python
x_values = np.linspace(0.1, 5, 1000)
y_values = [function_1(x) for x in x_values]
```

```python
plt.plot(x_values, y_values)
```

```python
# Gives a sequnce of values reaching the left hand limit of function "f" at x=a
def left_limit(f, a):
    x_values = np.linspace(a-0.01, a, 10, endpoint=False)
    y_values = [function_1(x) for x in x_values]
    return y_values
```

```python
def right_limit(f, a):
    x_values = np.linspace(a+0.01, a, 10, endpoint=False)
    y_values = [function_1(x) for x in x_values]
    return y_values
```

```python
right_limit(function_1, 3)
```

```python

left_limit(function_1, 2.5)
```

```python
function_1(3)
```
