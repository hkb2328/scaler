Why not 10000?
There is storage constraints.

1. We should not run out of inventory
2. We should not use too much space for a particular item

Let us stock upto 1300 units?

What is the probability of the stock getting over?

```python
1 - norm.cdf(1.5)
```

There are 200 stores, with similar sales data. How many stores will run out of inventory if we decide to stock up to 1300?

The "expected" number of stores which will run out of inventory is b

```python
200 * 0.066
```

What is the probability of 15 or more stores going out of inventory?

```python
1 - binom.cdf(n=200, p=0.06, k=14)
```

We want only 1% chance that each store will go out of inventory

```python
z = norm.ppf(0.99)
z
```

```python
1000 + z * 200
```

```python
1 - norm.cdf(z)
```
