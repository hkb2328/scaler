# Baseline Model

Now that we have the data ready,
- Lets create a simple 6 layered base model


|Layer| |Descrption|
|:--| |:--|
|**L1**| |Is the Input Layer |
|**L2**| |Contains 64 Neurons|
|**L3**| |Contains 32 Neurons|
|**L4**| |Contains 16 Neurons|
|**L5**| |Contains 8 Neurons |
|**L6**| |Is the Output Layer |

<br>

**Note:** In Between the layers, we will use:
- ReLU as the Activation function


````python
import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense


# For Reproducibility
np.random.seed(42)
tf.random.set_seed(42)
````

````python
def create_baseline():

  model = Sequential([
                    Dense(64, activation="relu",kernel_initializer='glorot_uniform'),
                    Dense(32, activation="relu",kernel_initializer='glorot_uniform'),
                    Dense(16, activation="relu",kernel_initializer='glorot_uniform'),
                    Dense(8, activation="relu",kernel_initializer='glorot_uniform'),
                    Dense(4,activation='softmax')])
  return model
````

````python
model = create_baseline()
````

Using ```Adam``` as Optimizer, and metrices as Precision and Recall
- Since Accuracy is not the right metric in Multi-Class setting  

````python
model.compile(optimizer = tf.keras.optimizers.Adam(),
                loss = tf.keras.losses.CategoricalCrossentropy(), metrics=['Precision','Recall']  )
````

Training the model with epoch=10 and batch size = 128

````python
history = model.fit(X_train, y_train, validation_data = (X_val, y_val),  epochs=10, batch_size=128)
````

````python
epochs = history.epoch
loss = history.history["loss"]
val_loss = history.history["val_loss"]


plt.plot(epochs, loss, label="train")
plt.plot(epochs, val_loss, label="validation")

plt.legend()
plt.title("Loss VS Epochs")

plt.show()

````

````python
epochs = history.epoch
precision = history.history["precision"]
val_precision = history.history["val_precision"]


plt.plot(epochs, precision, label="train")
plt.plot(epochs, val_precision, label="validation")

plt.legend()
plt.title("Precision VS Epochs")

plt.show()

````

````python
epochs = history.epoch
recall = history.history["recall"]
val_recall = history.history["val_recall"]


plt.plot(epochs, recall, label="train")
plt.plot(epochs, val_recall, label="validation")

plt.legend()
plt.title("Recall VS Epochs")

plt.show()

````

**Observe**

After 10 epochs:
- the model still having a high loss
- Along with that Precision is 68 %

The only thing the model improved upon:
- Is on the Recall which reached 35 %

#### What are the main hyperparameters to look at when using a NN ?

Ans: the main hyperparameters includes:

- Learning rate $\alpha$

- $β$ if using Gradient Descent with Momentum

- $β_1, β_2, ϵ$ when using Adam

- Number of layers of the NN

- Number of Neurons/ hidden units to be used

- Learning Rate decay

- Batch size  




Screenshot 2022-12-21 at 12.59.26 PM.png

#### But amongst all these hyperparameters, which one to tune first ?

Ans: Learning Rate $α$ ,
- since most of the time, due to an incorrect $α$
- the model fails to reach global minima

<br>

After the learning rate $α$, the next hyperparameters to tune should be :

- **$β$ value**: though a value of 0.9 is a good default for it

<br>

- **Number of neurons/hidden units**: As it helps in making the model more complex

<br>

- **Batch size**: Finding the right batch size, so that there is equal distribution of the data in every epoch


<br>

And finally the  hyperparameters which are last to tune will be :

- **Number of Layer of the NN** : As adding layers changes the NN entirely, it should be considered as the final resort

<br>

- **Learning Rate Decay**: If after getting the best learning rate, still the model does not perform, its a good practice to perform a decay of learning rate over epochs

<br>

- **$β_1, β_2, ϵ$ when using Adam**: Though not recommended to tune, but sometimes tuning these value help model converge to the global minima

#Gradient Checking

Screenshot 2023-02-09 at 6.52.34 PM.png

As we know the major issue of Neural Network is:
- It being a black box for interpretation

<br>

#### And Sometimes even after series of correct Hyperparameters, the Neural Network does not perform well, Why is that ?

Ans: Because of the gradients and backpropagation not performing correctly
- It can happen due to some bug/issue in gradient calculation

<br>

#### How to check if the gradients from backpropagation are correct ?

Ans: If we can approximate the gradient values and check
- Approximate gradient value $=$ Actual gradient value  

<br>

#### How to calculate Approximate gradient value ?

Ans: Lets take an example

Suppose we have a function $f(x) = x^3$, and the graph looks like the above snapshot
- Now if we take two points $p,q$ from a point $x=2$,such that:
 - $p = x + ϵ$
 - $q = x - ϵ$   
 - Where $ϵ = 0.01$

<br>

If we consider the triangle created by points $p,q$, then:
- the base of the traingle becomes: $p-q = x+ϵ - x+ϵ = 2ϵ$
- the height of the traingle becomes: $f(p)-f(q) = f(x+ϵ) - f(x-ϵ)$

<br>

And if we recall from calculas:
- The derivative of a function $f'(x)$ $= tan$ of the traingle created by the points on function
-  $tan = \frac{ f(x+ϵ) - f(x-ϵ)}{2ϵ}$










Screenshot 2023-02-09 at 6.52.41 PM.png

The value of $tan$ becomes:
- $tan = \frac{2.01^3 - 1.99^3}{2 \times 0.01} = \frac{8.120601 - 7.880599}{0.02} = 12.0001 $
<br>

The derivative of $f(x)$ is:
- $f'(x) = 3x^2 = 3(2)^2 = 12$




Hence we can see $\frac{f(x+ϵ) - f(x-ϵ)}{2ϵ} ≈ f'(x)$


**Note:** This is known as Gradient Checking

#### How to do gradient approximation in Neural Network ?

Ans: For backpropagation, every layer of Neural Network has:
- $ [W^1, b^1],[W^2, b^2], [W^3, b^3], ....., [W^L, b^L]  $
- Derivatives $ [ΔW^1, Δb^1],[ΔW^2, Δb^2], [ΔW^3, Δb^3], ....., [ΔW^L, Δb^L] $

- and loss is a function of $W^1, b^1,W^2, b^2,....W^L, b^L$
 - $Loss(W^1, b^1,W^2, b^2,....W^L, b^L)$

**Note:** Weight and derivates are 2-dimensional matrix

Screenshot 2023-02-09 at 6.52.50 PM.png



<br>

Lets create a single dimensional vector $\theta$ for every layer:
- By unrolling the weight and bias matrix and concatining it such that:
 - $ \theta_1 = [W^1, b^1] , \theta_2 = [W^2, b^2], \theta_3 = [W^3, b^3], ... \theta_L = [W^L, b^L]  $

<br>

Similarly derivatives become:

-  $ Δ\theta_1 = [ΔW^1, Δb^1] , Δ\theta_2 = [ΔW^2, Δb^2], Δ\theta_3 = [ΔW^3, Δb^3], ... Δ\theta_L = [ΔW^L, Δb^L]  $

<br>

- and loss becomes
 - $Loss(\theta_1, \theta_2 , ...,\theta_L )$

<br>

Now we concatinate all $\theta_1, \theta_2 , ...,\theta_L  $ into a single vector $θ$
- $θ = [\theta_1, \theta_2 , ...,\theta_L ]$

And loss becomes:
- $Loss(\theta) = Loss(\theta_1, \theta_2 , ...,\theta_L )$



#### How to implement gradient checking ?

Screenshot 2024-01-12 at 4.27.13 PM.png



Ans: Gradient checking is implemented as:

for i in range(0,len($θ$)):
-  $d\theta_{approx}[i] = \frac{L(\theta_1,\theta_2,\theta_3,...,[\theta_{i}+ϵ],..\theta_L)- L(\theta_1,\theta_2,\theta_3,...,[\theta_{i}-ϵ],..\theta_L)}{2 ϵ}$

and then check how close:
- $d\theta_{approx}[i] ≈ \frac{\partial L}{\partial θ_i}$

**Note** Value of $ϵ = 10^{-7}$



#### How to evaluate if the difference in approximation and actual gradient is correct or not ?


Ans: check value of $ \frac{||d\theta_{approx}[i] - Δθ_i||_2}{||d\theta_{approx}[i]||_2+||Δθ_i||_2} $ and if:

- value $\leq 10^{-5}$ there is no issue
- and if value $> 10^{-5}$ there is problem in Backpropagation



#### What to do if difference in approximation and actual gradient is really high ?

Ans: check for which weight $W^k$ , bias $b^k$ (where $k$ is the kth hidden layer) matrix is the difference high:
- then look for each weight/bias value to detect the bug

<br>


#### Should gradient checking be done during training ?

Ans: No, since gradient checking is time intensive,
- it can only be used for debugging purpose only


#### Does Gradient check work with dropout ?

Ans: No, because dropout drops neurons randomly in every iteration
- Which increases the complexity of calculating gradient of loss function










#Single Evaluation Metric

Screenshot 2022-12-19 at 2.13.40 PM.png

Notice that, in our model, we used two metrics Precision and Recall

- And the baseline model (Calling it Model A)
got Precision = 68% and Recall = 35%

<br>

If there is another model (Model B), such that:
- Precision = 64% but Recall = 40%

<br>

#### Can we say Model A performed better or Model B ?
Ans: No


Screenshot 2022-12-19 at 2.13.47 PM.png


#### How about we use F1-Score instead, will that help ?

Ans: Yes, as:
- $F1Score = \frac{2 Precision \times Recall}{Precision + Recall}$

Therefore
- F1 for modelA : $\frac{2 \times 0.68 \times 0.35}{0.68 + 0.35} = \frac{0.476}{1.03} = 0.462$

<br>

- And F1 for ModelB : $\frac{2 \times 0.64 \times 0.40}{0.64 + 0.40} = \frac{0.512}{1.04} = 0.492$


**observe**

How combining Precision and Recall in the form of F1Score,
- Shows that Model B performs better than Model A
- Even though Model A has higher Precision than Model B

<br>

**Note:** hence its always a good practice to use a single Evaluation Metric or combine metrices   


**quiz**

```
if there are 2 models with Precision and Recall as: M1 = [0.78, 0.88], M2 = [0.82, 0.67] then:

1. Model 1 performs better
2. Model 2 performs better

```

**answer**
1. Model 1 performs better

**Explanation**
As we have precision and Recall, we can find F1:

- $F1 = \frac{2\times Precision \times Recall}{Precision + Recall} $

therefore for Model 1,
- $M_1 = \frac{2 \times 0.78 \times 0.88 }{0.78 + 0.88}  = \frac{1.3728}{1.66} =0.83$

Similarly for Model 2,
- $M_2 = \frac{2 \times 0.82 \times 0.67 }{0.82 +0.67}  = \frac{1.0988}{1.49} = 0.74 $

Clearly Model 1 performs better




#### What if we cannot combine the metrices together ?

Ans: Lets take an example to understand it:

Suppose we have three models(A,B,C) , such that:


| Model | F1Score | Training Time |
| :-- | :--| :--|
|A| 80% | 100ms |
|B| 92% | 180ms |
|C|95%|250ms|

Now, we want a model
- which has maximum F1score and minimum Training time.

Clearly we cannot combine F1Score and Training Time together

<br>

Now what if we select the model:

- Which has the maximum F1Score
- Such that Training Time $\leq$ Threshold($\tau= 200$ ms)

<br>

#### Which will be the best model ?

Ans: Model B

<br>



#### Imagine if we have N Metrics which cannot be combined, then what to do ?

Ans: Consider 1 metric on which we want to the model to have maximum value in

- While conditioning the model on the remaining N-1 Metrices


**Note:** The metric:
- on which the model maximizes itself is called Optimizing metric
- while the metric/metrices on which conditions are applied are called Satisificing metric

Using F1 score as metrics

**Note:**

- The micro-averaged F1 score pools per-sample classifications across classes, and then calculates the overall F1 score.

- The macro-averaged F1 score computes a simple average of the F1 scores over classes.

<br>

Hence we want per-sample classification across all classes, hence using micro-averaged F1Score

````python
model = create_baseline()

f1 = tfa.metrics.F1Score(4,'micro')

model.compile(optimizer = tf.keras.optimizers.Adam(),
                loss = tf.keras.losses.CategoricalCrossentropy(), metrics=[f1]  )
````

training model for 10 epochs with batch size = 128

````python
history = model.fit(X_train, y_train, validation_data = (X_val, y_val),  epochs=10, batch_size=128)
````

````python
epochs = history.epoch
f1_score = history.history["f1_score"]
val_f1_score = history.history["val_f1_score"]


plt.plot(epochs, f1_score, label="train")
plt.plot(epochs, val_f1_score, label="validation")

plt.legend()
plt.title("F1Score VS Epochs")

plt.show()

````

Just by using F1Score, we can better understand that
- the model performance much clearly

#### Why compare to human level performance ?
Ans: As Humans tend to do many tasks effeciently and with precise accuracy
- For example classifying different brands of cars, voice recognition

- Also it helps the model Bias Variance tradeoff

<br>

#### how human level performance helps in Bias Variance of the model ?
Ans: Lets understand this with an example:

Supposedly we try to classify maruti cars from honda

- Now Human error is 1% while the model's training error is 7% and validation error is 9%

Now if we had only training and validation error,
- we would have said the model is overfitting meaning it has high variance
- and we should focus on variance

<br>

But due to the high gap between human error and model training error $(7-1)$% $= 6$%,
- Shows that the model is not fitting the training data well
- meaning the model is having a high bias
- showing that we should focus on bias

<br>

#### What if there are blur images such that Human error increases to 5.5 % and the model error remains same ?
Ans: As the training error and human error is quite close,
- it shows that the model is fitting well on the training data

But as validation error is more than training error,
- it means that the model has high variance

<br>

Hence the model should focus on Variance

<br>

**Note:** The difference/gap between Human and Training error is called **Avoidable Bias**
- Also for this example Human-level error is used as a proxy for Bayes optimal error

#### Why neural network works for complex real world problems ?

Ans: Suppose we are trying to create an autonomous vehicle, which in a particular frame checks:
- if the image consists of any Pedestrian, Car, Stop sign and Traffic Light

<br>

#### How will Machine learning do this task ?

Ans: Create 4 different Models which checks for Pedestrian, Car, Stop sign and Traffic Light as:
- Machine Learning is Sequential learning and can identify one object

**Note:** It has been seen classifying the image using 4 different models lead to a poor performance
- As no infromation is shared amongst the these Models

Screenshot 2023-02-09 at 6.52.24 PM.png



<br>

#### How Neural Network manages the issue of checking 4 objects in a single image ?

Ans: By creating an output layer with 4 neurons where each neuron represents an object
- and a value of 1 means the object is present and 0 meaning the object is not present

- These neurons share the features from the previous hidden layers and hence solves the issue of no information sharing
- which makes the model perform better


**Note:** This is known as Multi-Task Learning


#### Why no softmax is used ?

Ans: Softmax is useful when there is a need to identify one class over several classes
- But here we want to identify some classes over several classes
- Hence sigmoid is used with Crossentropy loss function
